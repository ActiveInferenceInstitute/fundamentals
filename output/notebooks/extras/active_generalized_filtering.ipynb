````python
from pathlib import Path
import runpy
import sys
import warnings

import matplotlib.pyplot as plt

# Locate the repository root at runtime (not the export machine's path):
# walk up from the notebook's working directory to the folder containing
# both `src/active_inference` and `chapters/`. This keeps a clean clone of
# the repo runnable regardless of who generated the notebook.
def _find_repo_root(start=None):
    here = Path(start or Path.cwd()).resolve()
    for candidate in [here, *here.parents]:
        if (candidate / 'src' / 'active_inference').is_dir() and (candidate / 'chapters').is_dir():
            return candidate
    return here

REPO_ROOT = _find_repo_root()
if str(REPO_ROOT / 'src') not in sys.path:
    sys.path.insert(0, str(REPO_ROOT / 'src'))

%matplotlib inline
warnings.filterwarnings('default')
plt.close('all')
````
